### Load Data

```python
from llama_index.core import SimpleDirectoryReader

# load documents
documents = SimpleDirectoryReader(input_files=["metagpt.pdf"]).load_data()
```

```python
from llama_index.core.node_parser import SentenceSplitter

splitter = SentenceSplitter(chunk_size=1024)
nodes = splitter.get_nodes_from_documents(documents)
```

### Define LLM & Embedding Model

```python
from llama_index.core import Settings
from llama_index.llms.openai import OpenAI
from llama_index.embeddings.openai import OpenAIEmbedding
```

```python
base_url = f'https://{spark.conf.get("spark.databricks.workspaceUrl")}/serving-endpoints'
databricks_token = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiToken().get()

Settings.llm = OpenAI(api_key=databricks_token, api_base=base_url, model="databricks-claude-sonnet-4")
Settings.embed_model = OpenAIEmbedding(api_key=databricks_token, api_base=base_url, model="databricks-gte-large-en")
```
